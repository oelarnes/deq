```python
from ipy import *
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import plotly


%load_ext autoreload
%autoreload 2
```

```python
ax = plt.subplot()
metrics = [
    "deq_strat_delta",
    "gih_wr_strat_delta",
    "gp_wr_strat_
    delta",
    "pick_equity_strat_delta",    
    "iwd_strat_delta",  
]
for metric in metrics:
    ax.plot("wr_group", 
            metric, 
            data=res.df)
    

ax.legend()
ax.yaxis.set_major_formatter(
    mtick.PercentFormatter(1.0
                          ))
ax.xaxis.set_major_formatter(
    mtick.PercentFormatter(100))


ax.set_title("Effect on win rate of choosing p1p1 by metric")
ax.set_xlabel('Skill Cohort')
ax.set_ylabel('Simulated effect on match win rate')

plt.show()
```
