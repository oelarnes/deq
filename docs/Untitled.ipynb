```python
import polars as pl
from spells import summon, ColSpec, ColType, ColName

ext = {
    'num_creatures': ColSpec(
        col_type=ColType.NAME_SUM,
        expr=lambda name, card_context: pl.col(f"deck_{name}")
        * (1 if "Creature" in card_context[name][ColName.CARD_TYPE] else 0)
    ),
    'deck_num_creatures': ColSpec(
        col_type=ColType.GROUP_BY,
        expr=lambda names: pl.sum_horizontal([pl.col(f"num_creatures_{name}") for name in names])
    )
}

%time summon('OTJ', columns=['game_wr', 'num_games'], group_by=['deck_num_creatures', 'main_colors'], extensions=ext, filter_spec={'player_cohort': 'Middle'}).filter(pl.col('num_games') > 100)
```
