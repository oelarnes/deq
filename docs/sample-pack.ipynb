```python
from spells import summon
import polars as pl

from deq import ext
ata_v_gp = summon(["OTJ", "BLB", "FDN"]
                  
                  , ["ata", "gp_wr_17l"],
                  filter_spec={'format_day': 3},
                 extensions=ext).filter(
    pl.col("gp_wr_17l").is_not_null())
```

```python
import matplotlib.pyplot as plt
_, ax = plt.subplots()

x = ata_v_gp["ata"]
y = ata_v_gp["gp_wr_17l"]
ax.scatter(x, y)

import numpy as np
m, b = np.polyfit(x, y, 1)
ax.plot(x, m*x+b, color="red")
print(m)

yr = y - m*x + b
np.sqrt(yr.var())
```
