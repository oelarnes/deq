```python
metric_filter = {'lhs': 'format_day', 'op': '<=', 'rhs': 14}
context = deq_bias_set_context(
    'OTJ', 
    metric_filter,
    observed_days=14
)
df = summon('OTJ', ['deq', 'ata', 'num_gp', 'gp_wr', 'gp_wr_b', 'gp_wr_mean', 'pct_gp', 'pick_equity', 'deq_base',
                   'gp_wr_bias', 'deq_bias_adj', 'deq_meta_adj', 'meta_regression_factor'], 
            group_by=['name', 'expansion'], filter_spec = metric_filter, extensions=ext, set_context=context).with_columns(
    pl.col('gp_wr_b').alias('mwr') - pl.col('gp_wr_mean'))
df.filter(~pl.col('deq').is_null()).sort('deq', descending=True)
```
